## Euclidean Algorithm 
---

````python
def euclidean_gcd(a: int, b: int) -> int:
    """
    This function implements the euclidean algorithms
    to get the greater common divisor of two numbers

    Args:
        a(int): First number in the algorithm. Expected to be greater than the other
        b(int): Second number in the algorithm.
    
    Returns:
        int: The greater common divisor of the two numbers
    """
    if b>a:
        a, b = b, a
    while b!= 0:
        r = a%b
        a = b
        b = r
    return a

print(euclidean_gcd(36, 54))
````

````python
resultado = euclidean_gcd(24, 60)
````

## Babylonean Square Root Method
---

````python
def square_root_babylonean_method(s: int) -> float:
    """
    This function implements the babylonean method to get the approximate square root of a number.

    Args: 
        s(int): Input number to calculate the square root.
        
    Returns:
        A decimal number that is the approximate square root of the input number.
    """
    x_n = s/2
    diff = 0.0001
    x_n_1 = (1/2)*(x_n + s/x_n)
    while abs(x_n_1 - x_n) > diff:
        x_n = x_n_1
        x_n_1 = (1/2)*(x_n + s/x_n)
    return x_n_1
````

````python
resultado = square_root_babylonean_method(25)
print(resultado)
resultado_1= euclidean_gcd(24, 60)
print(resultado_1 + resultado)
````

# Maximum element in an array
---

````python
def maximum_array(array: list) -> int:
    """
    This function gets the maximum value into a given array.

    Args:
        array(list): Array of numbers

    Returns:
        An integer number with the maximum in the array. If array is empty,
        a "None" value will be returned.
    """
    if len(array) > 0:
        max = array[0]
        for number in array:
            if number > max:
                max = number    
    else:
        #array is empty
        max = None

    return max     #Tener un solo valor por funcion, no en cada condicional
````

````python
arreglo = [1, 2, 3, -3, 99]
print(maximum_array(arreglo))
````

````python
def sumatory(n: int) -> int:
    """
    This function calculates the sum of the squares of the first n natural numbers.
    
    Args:
       n(int): The number of natural numbers to consider.
    
    Returns:
        int: The sum of the squares of the first n natural numbers.
    """
    sums = 0
    for i in range(1,n+1): #Ya que el rango es hasta n-1, se le suma 1 para que llegue a n
        sums += i**2
    return sums
n = int(input("Enter a number: "))
formula = int((n*(n+1)*(2*n+1))/6) #Las operaciones con división devuelcen el resultadfo con decimales.
print(f'Sumatoria: {(sumatory(n))}\tFormula: {formula}')
if formula == sumatory(n):
    print(True)
else:
    print(False)
````

# Factorial
---

````python
def factorial(n: int) -> int:
    """Calculates the factorial of a given number."""
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)
def factorial_2(n: int) -> int:
    "This function calculates the factorial of a given number using an iterative approach." 
    result = 1
    for i in range(1, n+1 ):
        result *= i
    return result
n = int(input("Enter a number to calculatre its factorial: "))
print(f'Mathematical: {factorial(n)}\tIterative: {factorial_2(n)}')
````

## Brute Force
---
**Concept:** try all possible solutions until finding the correcty one
Characteristics: 
* Exhaustive: Examines every posibility 
* Guaranteed: Always finds optimal solution(if exists)
* Expensive: Often exponential time complexity
* Simple: Easy to understand and implement  
**When to use it**
    * Problem size is small 
    * No efficient algorithm is known 
    * Correctness is more important than efficiency
    * As baseline for comparing other algorithms
    * When optimization overhead exceeds brute force cost

## Greedy Algorithm: Philosophy and Approach
Definition: A greedy algorithm makes locally optimal choices at each step, hoping to find a global optimum
"Algoritmos glotones o voráces
### Strategy
1. At each step, choices the best available option
2. Optimal solutions contains optimal 
3. Not tracker



````python
#solve with greedy algorithm
#Given: Coin denomnations [d1, d2, ..., dk] and amount n
# Find minimum number of coins to make amount n
def greedy_coins_iterative(denominations : list, amount: int) -> list:
    denominations.sort(reverse=True) #Voltear la lista
    #o sorted(denominations, reverse = True)
    result = [] #monedas
    for d in denominations: #para cada denominación en la lista 
        while amount >= d:
            result.append(d)
            amount -= d
            #añadir el número de monedas de esta denominación
         #actualizar el monto restante
    return result, len(result) #retornar la lista de monedas y el número de monedas
def greedy_coins_maths(denominations: list, amount: int)-> list:
    denominations = sorted(denominations, reverse = True)
    result = []
    for d in denominations: 
        result. append(d) #CORREGIR
        amount = amount%d
    return result


coins = [1, 5, 10, 25]
amount = 45
print(greedy_coins_iterative(coins, amount))
print(greedy_coins_maths(coins, amount))
````
