### Brute Force
---
**Concept:** try all possible solutions until finding the correcty one
Characteristics: 
* Exhaustive: Examines every posibility 
* Guaranteed: Always finds optimal solution(if exists)
* Expensive: Often exponential time complexity
* Simple: Easy to understand and implement  
**When to use it**
    * Problem size is small 
    * No efficient algorithm is known 
    * Correctness is more important than efficiency
    * As baseline for comparing other algorithms
    * When optimization overhead exceeds brute force cost

# Backtracking
---
> Systemic Trial and Error

* Definition: explores solution space systematically, abandoning __partial solutions__ that cannot lead to valid solutions
1. Choose: Make a choice from available options
2. Explore: Recursively explore consequences
3. Unchoose: If path fails, backtrack and try another (Si parece que no va a funcionar, se devuelve y usar otra estrategia)

```python
Algorithm BACKTRACK (partial_solution)
if IsComplete(partial_solution) then
    return partial_solution
end if
for each possible choice c do:
    if IsValid(partial_solution, c) then
        partial_solution.add(c)
        result <- BACKTRACK(partial_solution)
        if result != null then
            return result
        end if
        partial_solution.remove(c) //Backtrack
    end if
end for
return null
```

## Interpolation Search
---
Interpolation search improve upon binary search ny making educated guesses about where the target element might be located, based on the values at the endpoints.

>Key Concept
Instead of always choosing the middle element, interpolation search estimates the position using:
$$
pos = low+ \frac{((x-A[low])}{(A[high])-A[low])} * (high-low)
$$

## Search Algorithm Selection Criteria
---
* Data Size: How many elements to search?
* Data Organization: Is the data sorted?
* Search Frequency: One-Time or repeated searches
* Data Distribution: Uniformly distributed values?
* Memory Constrains:

````python
# Linear
def Linear_search_guess(numbers: list, number: int)-> int:
    index = -1
    for i, element in numbers:
        if element == number:
            index = i
            break

    return index
````

````python
# Recursivo
def binary_search_recursive_guess(a:list, x: int):
    if a==[]:
        return False
    low = 0
    high = len(a)-1
    while low <= high:
        mid = (low+high)//2
        if a[mid] == x:
            return True
        elif a[mid] < x:
            return binary_search_recursive_guess(a[mid + 1:], x) 
        else:
            return binary_search_recursive_guess(a[:mid], x) 
    return False
````

## Greedy Algorithm: Philosophy and Approach
Definition: A greedy algorithm makes locally optimal choices at each step, hoping to find a global optimum
"Algoritmos glotones o voráces
### Strategy
1. At each step, choices the best available option
2. Optimal solutions contains optimal 
3. Not tracker



````python
#solve with greedy algorithm
#Given: Coin denomnations [d1, d2, ..., dk] and amount n
# Find minimum number of coins to make amount n
def greedy_coins_iterative(denominations : list, amount: int) -> list:
    denominations.sort(reverse=True) #Voltear la lista
    #o sorted(denominations, reverse = True)
    result = [] #monedas
    for d in denominations: #para cada denominación en la lista 
        while amount >= d:
            result.append(d)
            amount -= d
            #añadir el número de monedas de esta denominación
         #actualizar el monto restante
    return result, len(result) #retornar la lista de monedas y el número de monedas
def greedy_coins_maths(denominations: list, amount: int)-> list:
    denominations = sorted(denominations, reverse = True)
    result = []
    for d in denominations: 
        result. append(d) #CORREGIR
        amount = amount%d
    return result


coins = [1, 5, 10, 25]
amount = 45
print(greedy_coins_iterative(coins, amount))
print(greedy_coins_maths(coins, amount))
````

````python
import sys
import math
from itertools import groupby

# Auto-generated code below aims at helping you parse
# the standard input according to the problem statement.

#message = input()
message = input()

mensaje_binario = "".join(f"{ord(char):07b}" for char in message)
#Crea una cadena completa sin espacios con
#cada caracter del mensaje, lo cambia a codigo ASCII, lo pasa a binario 
#lo deja en formato de 7 bits unicamente y lo añade a la cadena

unarios = [] #Lista para agrupar cada caracter en unario
for bit, grupo_bits in groupby(mensaje_binario): #Agrupar por N° de bits consecutivos iguales
    cantidad_bits = len(list(grupo_bits)) #Cuenta la cantidad de bits en un solo grupo
    prefijo = "0" if bit == "1" else "00" #Para empezar con 1 o 0 en unario
    ceros = "0" * cantidad_bits #Para la cantidad de digitos 1s o 0s
    unarios.append(f"{prefijo} {ceros}") #Añade un elemento a la lista con 
    #el prefijo y los ceros con espacio intermedio

# 3. Imprimir el resultado uniendo todos los bloques con un espacio
print(" ".join(unarios)) #Join añade un espacio en medio de cada elemento de la lista
# Write an answer using print
# To debug: print("Debug messages...", file=sys.stderr, flush=True)
````
