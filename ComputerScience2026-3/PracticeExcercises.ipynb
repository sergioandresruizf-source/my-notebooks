## Euclidean Algorithm 
---

```python
#Euclidean Algorithm
def euclidean_gcd(a: int, b: int) -> int:
    """
    This function implements the euclidean algorithms
    to get the greater common divisor of two numbers

    Args:
        a(int): First number in the algorithm. Expected to be greater than the other
        b(int): Second number in the algorithm.
    
    Returns:
        int: The greater common divisor of the two numbers
    """
    if b>a:
        a, b = b, a
    while b!= 0:
        r = a%b
        a = b
        b = r
    return a

print(euclidean_gcd(36, 54))
```

## Babylonean Square Root Method
---

```python
def square_root_babylonean_method(s: int) -> float:
    """
    This function implements the babylonean method to get the approximate square root of a number.

    Args: 
        s(int): Input number to calculate the square root.
        
    Returns:
        A decimal number that is the approximate square root of the input number.
    """
    x_n = s/2
    diff = 0.0001
    x_n_1 = (1/2)*(x_n + s/x_n)
    while abs(x_n_1 - x_n) > diff:
        x_n = x_n_1
        x_n_1 = (1/2)*(x_n + s/x_n)
    return x_n_1
```

# Maximum element in an array
---

```python
def maximum_array(array: list) -> int:
    """
    This function gets the maximum value into a given array.

    Args:
        array(list): Array of numbers

    Returns:
        An integer number with the maximum in the array. If array is empty,
        a "None" value will be returned.
    """
    if len(array) > 0:
        max = array[0]
        for number in array:
            if number > max:
                max = number    
    else:
        #array is empty
        max = None

    return max     #Tener un solo valor por funcion, no en cada condicional
```

```python
def factorial(n: int) -> int:
    """Calculates the factorial of a given number."""
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)
def factorial_2(n: int) -> int:
    "This function calculates the factorial of a given number using an iterative approach." 
    result = 1
    for i in range(1, n+1 ):
        result *= i
    return result
n = int(input("Enter a number to calculatre its factorial: "))
print(f'Mathematical: {factorial(n)}\tIterative: {factorial_2(n)}')
```

## N-Queens
---

```python
def is_safe(board: list, row: int, col: int, n: int)->bool:
    answer = True
    #no queeens in the same column
    for i in range(row):
        if board[i][col] == "Q":
            #   There's a queen
            answer = False
            break
    #no queens in the same diagonal
    for i in range(row):
        n_r = row-i-1
        n_c = col-i-1
        if (n_r >= 0 and n_c >= 0) and (board[n_r][n_c] == "Q"):
            answer = False
            break
        n_c = col+i+1
        if (n_r >= 0 and n_c < n ) and (board[n_r][n_c] == "Q"):
            answer = False
            break

    return answer

def NQueens(board: list, row: int, n: int) -> int:
    '''This algorithm analize a number n of queens that need to be positioned in a board nxn
    
    Args: 
        board(list): 
        row(int):
    
    Returns:
        result(board, row +1):
    '''
    if row == n:
        return board #No found solution
    for c in range(n):
        if is_safe(board, row, c, n):
            board[row][c] = "Q"
            solution = NQueens(board, row+1, n) 
            if solution is not None:
                return solution
            board[row][c] = ""
    return None

n = int(input())
row = 0
board = [['' for _ in range(n)] for __ in range(n)]

solution = NQueens(board, 0, n)
for row in solution:
    print(row)

            



```

## Binary Search
---
Binary search is a divide and conquer algorithm that finds an element in a sorted array by repeatdly dividing a list in less steps than linear search.

```python
def binary_search(a:list, x: int):
    if a==[]:
        return False
    low = 0
    high = len(a)-1
    while low <= high:
        mid = (low+high)//2
        if a[mid] == x:
            return mid
        elif a[mid] < x:
            low = mid + 1
        else:
            high = mid -1
    return -1
```

```python
from random import randint
x = randint(1, 1000001)
listona = [i for i in range(1, 1000001)]
from time import time
start = time()
index = binary_search(listona, x)
end = time()
print(f'El elemento {x} está en {index} y de {time(end - start)}') #CORREGIR EN LA CASA
```

## Forma recursiva
---


```python
def binary_search_recursive(a:list, x: int):
    if a==[]:
        return False
    low = 0
    high = len(a)-1
    while low <= high:
        mid = (low+high)//2
        if a[mid] == x:
            return True
        elif a[mid] < x:
            return binary_search_recursive(a[mid + 1:], x) 
        else:
            return binary_search_recursive(a[:mid], x) 
    return False
```

```python
from random import randint
array = [i for i in range(1,1000001)]
x = randint(1, 1000001)
print(f'Con recursivo da {binary_search_recursive(array, x)}')
```

## Añadir el código de interpolation
---

```python
def busqueda_interpolacion(A, x):
    low = 0
    high = len(A) - 1

    # El arreglo debe estar ordenado y el elemento dentro del rango potencial
    while low <= high and A[low] <= x <= A[high]:
        # Si los extremos son iguales, evitamos división por cero
        if A[low] == A[high]:
            if A[low] == x:
                return low
            break
        
        # Fórmula exacta de la imagen para calcular la posición estimada
        pos = low + int(((x - A[low]) / (A[high] - A[low])) * (high - low))

        # Si encontramos el elemento objetivo
        if A[pos] == x:
            return pos

        # Si el elemento es mayor, está en la mitad superior
        if A[pos] < x:
            low = pos + 1
        # Si el elemento es menor, está en la mitad inferior
        else:
            high = pos - 1

    return -1  # Retorna -1 si el elemento no se encuentra

# --- Ejemplo de uso ---
arreglo = [10, 12, 13, 16, 18, 19, 20, 21, 22, 23, 24, 33, 35, 42, 47]
target = 18

indice = busqueda_interpolacion(arreglo, target)
print(f"Elemento encontrado en el índice: {indice}")  # Output: 4
```

## Bubble sort, Insertion sort and Selection sort
---

```python

#bubble sort
def bubblesort(A: list)->list:

    n = len(A)
    for i in range(n-1):
        for j in range(n - 1 - i):
            if A[j] > A[j+1]:
                swap = A[j]
                A[j] = A[j+1]
                A[j+1] = swap
    return A
```

```python
from random import randint
A = [randint(1,1000) for _ in range(100)]
print(A)
A = bubblesort(A)
print(A)
```

```python
#selection
def selection_sort( A:list)-> list:
    n = len(A)
    for i in range(n - 1):
        min_index = i
        for j in range(i + 1, n):
            if A[j] < A[min_index]:
                min_index = j
        swap = A[i]
        A[i] = A[min_index]
        A[min_index] = swap
    return A
```

```python
from random import randint
A = [randint(1,1000) for _ in range(100)]
print(A)
A = selection_sort(A)
print(A)
```

```python
lista_notas= [('Diane', 2.1), ('Charlie', 3.9), ('Bob', 2.5), ('Eve', 3.7), ('Diana', 3.8)]
def selection_sort_students( A:list)-> list:
    n = len(A)
    for i in range(n - 1):
        max_index = i
        for j in range(i + 1, n):
            if A[j][1] > A[max_index][1]:
                max_index = j
        swap = A[i]
        A[i] = A[max_index]
        A[max_index] = swap
    return A

print(selection_sort_students(lista_notas))
```

```python
#insertion sort
def insertion_sort(A:list)->list:
    n = len(A)
    for i in range(2, n):
        key = A[i]
        j = i - 1
        while j >= 1 and A[j] > key:
            A[j + 1] = A[j]
            j -= 1
        A[j + 1] = key

#COMPLETAR Y PROBAR
```
