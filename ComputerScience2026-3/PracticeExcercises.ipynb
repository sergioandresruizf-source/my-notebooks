# Optical Reader
---

```python
n = int(input("N: "))
letters = ["A", "B", "C", "D", "E"]
while n != 0:
    for i in range(n):
        answers = list(map(int, input("Respuestas: ").split(" ")))
        #answers_int = list(map(int, answers))
        output = "*"
        flag = False
        for i, answers in enumerate(answers):
            if answers <= 127:
                if not flag:
                    output = letters[i]
                    flag = True
                else:
                    output = "*"
                    break
        print(output)        
    n = int(input("N:"))
```

```python
numbers = input("Enter x1, y1, x2, y2 (or '0 0 0 0' to exit): ")
while numbers != ("0 0 0 0"):
    x1, y1, x2, y2 = tuple(map(int, numbers.split(" ")))
    output = 2
    if x1 == x2:
        if y1 == y2:
            output = 0
        else: 
            output =1 #Same Row
    elif y1 == y2:
        output = 1
    elif abs(x1-x2) == abs(y1-y2):
        output = 1
    print(output)
    numbers = input()
```
