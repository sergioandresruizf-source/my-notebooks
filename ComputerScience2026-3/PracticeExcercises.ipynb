```python
denominations = [50000, 20000, 10000, 2000, 500, 200, 100, 50]
denominations = sorted(denominations, reverse=True)
denominations.append(200000)
denominations.sort(reverse=True)
print(denominations)
print(denominations[0:9:2])
print(denominations[2:8])
print(len(denominations))
denominations.insert(3, 5000)
denominations.append([4,5])
denominations[denominations.index(2000)] = 2500
print(denominations)
```
